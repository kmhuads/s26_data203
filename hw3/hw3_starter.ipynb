````python
### FUNCTION TO STUDY AND DESCRIBE ###
def fetch_openaq_data(key, lat, lon, outfile):    
    from openaq import OpenAQ
    import json 
    
    client = OpenAQ(api_key=key)    
    locations = client.locations.list(
        coordinates=(lat, lon),
        radius=20_000,
        limit=1000
    )
    json.dump(locations.dict()['results'], open(outfile, "w", encoding="utf-8")) 
    client.close()
````

````python
def build_location_table(pa_sensor, openaq_location_dataset):
    
    import pandas as pd
    import json 
    
    df = pd.DataFrame()
    df = pd.concat([df, pd.DataFrame().from_dict(pa_sensor, orient='index')],axis=1)
    
    for r in openaq_location_dataset:
        d = r['coordinates']
        d['name'] = r['name']
        d['owner'] =  r['instruments'][0]['name']
        df = pd.concat([df, pd.DataFrame().from_dict(d, orient='index')], axis=1)
    
    df = df.T.reset_index(drop=True)
    
    return df
````

returns **all** the reference station's distances to index `0` or

````python
### STUDY THIS FUNCTION -> RUN IT TOO!!!
def n_closest(s1, df, df_dist_tbl, n=1, reference_only=False):
    if reference_only:
        df_ref_idx = df.query('owner == "Government Monitor"').index
        d = df_dist_tbl \
            .iloc[df_ref_idx, s1] \
            .sort_values(ascending=True) \
            .take(range(1, len(df_ref_idx) if n < 0 or n >= len(df_ref_idx) else n+1))
    else:
        d = df_dist_tbl \
            .iloc[:,s1] \
            .sort_values(ascending=True) \
            .take(range(1, n+1))
    
    return list(zip(d.index, d.values))
````
